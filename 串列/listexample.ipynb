# 簡介
串列是一個可以儲存多元素(整數,浮點數,或字串)的容器

## 索引值

```python
scores = [78, 29, 56, 80, 99]
print(scores[0])
print(scores[1])
```

## 串列內容的個數
len(a)取得a這個列串的個數

```python
scores = [78, 29, 56, 80, 99]
print(scores[0])
print(scores[1])
len(scores)
print(scores[len(scores)-1])
print(scores[-1])
```

## 新增元素

使用`append()`

```python
scores = [78, 29, 56, 80, 99]
print(scores[0])
print(scores[1])
len(scores)
print(scores[len(scores)-1])
print(scores[-1])
scores.append(71)
```

```python
weights = [12]*6
print(weights)
```

```python
scores = [78, 29, 56, 80, 99]
scores.append(71)
print(scores)
```

## 走訪串列的三種寫法

```python
print(scores)
for i in range(len(scores)):
    print(scores[i])
```

```python
#對於 scores 中的每個元素,一次取一個,放在e中
for e in scores:
    print(e)
```

```python
for i, e in enumerate(scores):
    print(i, e)
```

```python
fid = [0]*91
print(fid)
fid[0] = 1
fid[1] = 1
print(fid)
fid[2] = fid[0] + fid[1]
print(fid)
fid[3] = fid[1] + fid[2]
print(fid)
for i in range(2,91):
    fid[i] = fid[i-1] +fid[i-2]
print(fid[90])
```
