```python
import pyautogui as pyg

# 開啟 PyAutoGUI 的滑鼠安全防護（把滑鼠移到螢幕左上角 (0,0) 即可強制終止程式）
pyg.FAILSAFE = True

print("請將遊戲視窗準備好，3 秒後開始...")
pyg.sleep(3)

# 模擬按下空白鍵開始遊戲
pyg.press('space')

print("自動遊戲中... (若要強制停止，請將滑鼠移到螢幕左上角)")

try:
    while True:
        # 抓取畫面
        im = pyg.screenshot()
        obs = False
        
        # 檢查障礙物偵測區域（可依你的螢幕解析度與瀏覽器縮放微調座標）
        for i in range(15):
            # 檢查前方像素顏色（支援常見的淺灰/深灰色障礙物）
            c = im.getpixel((139 + i, 437))
            
            # 偵測障礙物顏色 (Chrome 恐龍的障礙物通常是灰色系)
            if c == (83, 83, 83) or c == (172, 172, 172) or c == (53, 53, 53):
                obs = True
                break
                
        # 如果偵測到障礙物，立刻按下空白鍵跳躍
        if obs:
            pyg.press('space')
            
except KeyboardInterrupt:
    print("\n手動終止程式。")
```
