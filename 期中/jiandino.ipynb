```python
import time
import ctypes
from tkinter import Tk
from PIL import ImageGrab  # 如果沒有 PIL，下方提供完全 zero-dependency 的 win32 擷取

# 使用 Windows API 模擬按鍵 (完全不需 pip install pyautogui)
VK_SPACE = 0x20
KEYEVENTF_KEYUP = 0x0002

def press_space():
    """使用 Windows 底層 API 模擬按下與放開空白鍵"""
    ctypes.windll.user32.keybd_event(VK_SPACE, 0, 0, 0)         # 按下 Space
    time.sleep(0.05)
    ctypes.windll.user32.keybd_event(VK_SPACE, 0, KEYEVENTF_KEYUP, 0) # 釋放 Space

print("==========================================")
print("程式將在 3 秒後開始，請立即切換至小恐龍遊戲畫面！")
print("想結束程式請在 VS Code 終端機按下 Ctrl + C")
print("==========================================")
time.sleep(3)

# 發送第一次空白鍵開始遊戲
press_space()

# 設定偵測點 (請根據螢幕解析度調整)
# 以 1920x1080 視窗最大化為例
DETECT_X = 100                   # 小恐龍前方的 X 座標
DETECT_Y = 360  # 仙人掌高度的 Y 座標

# 取得螢幕 DC 處理畫素 (完全不需 Pillow 套件)
user32 = ctypes.windll.user32
gdi32 = ctypes.windll.gdi32
hdc = user32.GetDC(0)

try:
    while True:
        # 取得指定座標的 RGB 顏色 (COLORREF 格式: 0x00BBGGRR)
        color = gdi32.GetPixel(hdc, DETECT_X, DETECT_Y)
        
        # 提取 Red 通道數值
        r = color & 0xFF
        
        # 當偵測到深色障礙物 (RGB 數值較低，代表樹枝或翼龍靠近)
        if r < 100:
            press_space()
            time.sleep(0.1)  # 防連點冷卻時間

except KeyboardInterrupt:
    print("\n程式已停止。")
finally:
    user32.ReleaseDC(0, hdc)
```

```python
import time
import ctypes

user32 = ctypes.windll.user32
gdi32 = ctypes.windll.gdi32

print("請在 3 秒內將滑鼠游標移動到【小恐龍前方約 5~10 公分的仙人掌高度】...")
time.sleep(3)

class POINT(ctypes.Structure):
    _fields_ = [("x", ctypes.c_long), ("y", ctypes.c_long)]

pt = POINT()
user32.GetCursorPos(ctypes.byref(pt))
hdc = user32.GetDC(0)
color = gdi32.GetPixel(hdc, pt.x, pt.y)
user32.ReleaseDC(0, hdc)

print(f"你的精準座標為 -> DETECT_X = {pt.x}, DETECT_Y = {pt.y}")
print(f"該點目前顏色 R Channel: {color & 0xFF}")
```

```python
import time
import ctypes

# Windows Virtual Key Codes
VK_SPACE = 0x20
VK_DOWN = 0x28
KEYEVENTF_KEYUP = 0x0002

user32 = ctypes.windll.user32
gdi32 = ctypes.windll.gdi32

def press_key(vk_code):
    """模擬按下與釋放指定按鍵"""
    user32.keybd_event(vk_code, 0, 0, 0)
    time.sleep(0.03)
    user32.keybd_event(vk_code, 0, KEYEVENTF_KEYUP, 0)

def fast_drop():
    """模擬發送向下鍵急速下墜"""
    user32.keybd_event(VK_DOWN, 0, 0, 0)
    time.sleep(0.08)  # 強制按住一小段時間讓角色迅速落地
    user32.keybd_event(VK_DOWN, 0, KEYEVENTF_KEYUP, 0)

# 請設定兩個偵測點座標：
# DETECT_X1 / Y1: 第一起跳點 (較靠近恐龍)
# DETECT_X2 / Y2: 預見第二起跳點 (較遠處，用來判斷是否連在一起)
DETECT_X1, DETECT_Y1 = 465,410
DETECT_X2, DETECT_Y2 = 354,410
 
hdc = user32.GetDC(0)
print("3 秒後開始執行『自動起跳 + 急速降落』邏輯...")
time.sleep(3)
 
press_key(VK_SPACE)  # 開始遊戲
 
try:
    while True:
        # 讀取兩個偵測點的顏色 R Channel
        color1 = gdi32.GetPixel(hdc, DETECT_X1, DETECT_Y1) & 0xFF
        color2 = gdi32.GetPixel(hdc, DETECT_X2, DETECT_Y2) & 0xFF

        # 1. 偵測到第一顆仙人掌 -> 起跳
        if color1 < 100:
            press_key(VK_SPACE)

            # 2. 起跳後立刻檢查第二點：如果第二點也有仙人掌 (間距過近)
            if color2 < 100:
                time.sleep(0.12)  # 等待完全飛過第一顆仙人掌的頂端
                fast_drop()        # 執行急速下墜！
                
            time.sleep(0.1)  # 防連點冷卻

except KeyboardInterrupt:
    print("\n已停止程式")
finally:
    user32.ReleaseDC(0, hdc)
```
