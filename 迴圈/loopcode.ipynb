# 九九乘法表
1\*1=1,1\*2=2,1\*3=3,,,,,,  
2\*1=2,2\*2=4,2\*3=6,,,,,,  

```python
for j in range(1,6)
j=1
print(f'1*{j}={1*j}', end=',')
j=2
print(f'1*{j}={1*j}', end=',')
j=3
print(f'1*{j}={1*j}', end=',')
```

```python
i=1
for j in range(1,6):
     print(f'{i}*{j}={i*j}', end=',')
print()
i-2
for j in range(1,6):
     print(f'{2}*{j}={i*j}', end=',')
print()
```

# 練習
將上面每一row最後面的逗號移除

```python
i=1
for j in range(1,6):
     print(f'{i}*{j}={i*j}', end=',')
print()
i-2
for j in range(1,6):
     print(f'{i}*{j}={i*j}', end=',')
print()
i=3
for j in range(1,6):
     print(f'{i}*{j}={i*j}', end=',')
print()
```
