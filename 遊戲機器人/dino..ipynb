# 介紹
使用 pyautogui 來完成小恐龍的遊戲機器人。

小恐龍是在 chrome 中使用 chrome://dino 啟動。

邏輯：

不斷監視畫面
讀取某個位置的顏色
若此位置上出現 (83, 83, 83) 的顏色
跳起來


```python
import  pyautogui as pyg
pyg.mouseInfo()
pyg.sleep(2)
pyg.press('space')
while True:
    im = pyg.screenshot()
    obs = False
    for i in range(13):
       c = im.getpixel((135+i, 424))
       if c ==(83, 83, 83):
           obs = True
           break
       if obs == True:
           pyg.press('space')
```
